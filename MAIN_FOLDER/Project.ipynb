## Inspect XML Annotations and Identify Road-Damage Classes

### Objective
The objective of this step is to examine the XML annotation files associated with the road-damage dataset. These annotations contain information about the image filenames, damage categories, and bounding-box coordinates required for training an object detection model such as YOLO.

### Methodology
1. **Locate the dataset:** Define the paths to the road-damage dataset, training images, and XML annotation directory using Python's `pathlib` module.
2. **Count annotation files:** Identify the number of XML files available for the training dataset.
3. **Extract damage categories:** Parse each XML file using `xml.etree.ElementTree` and count the annotated objects belonging to each damage class.
4. **Verify image availability:** Check whether the image filenames recorded in the XML annotations correspond to files present in the training image directory.
5. **Count annotated objects:** Calculate the total number of road-damage instances represented by the bounding-box annotations.
6. **Inspect a sample annotation:** Display part of an XML file to understand its structure and identify the information needed for later conversion into YOLO format.

### Libraries Used
- **Pathlib:** Locates dataset folders and files.
- **xml.etree.ElementTree:** Reads and parses XML annotation files.
- **collections.Counter:** Counts the occurrences of each road-damage category.

### Expected Outcome
This inspection will help us understand the dataset's annotation structure

```python

from pathlib import Path
import xml.etree.ElementTree as ET
from collections import Counter

dataset = Path("Road_Damage_Dataset")

xml_dir = dataset / "labels/train_labels/xmls"
train_dir = dataset / "images/train_images"

xml_files = list(xml_dir.glob("*.xml"))

print("Total XML files:", len(xml_files))
print("Training image folder exists:", train_dir.exists())

class_counts = Counter()
missing_images = []
total_boxes = 0

for xml_file in xml_files:
    try:
        tree = ET.parse(xml_file)
        root = tree.getroot()

        filename = root.findtext("filename")

        if filename:
            image_path = train_dir / filename
            if not image_path.exists():
                missing_images.append(filename)

        for obj in root.findall("object"):
            name = obj.findtext("name", default="Unknown").strip()
            class_counts[name] += 1
            total_boxes += 1

    except ET.ParseError:
        print("Invalid XML:", xml_file.name)

print("\nDamage classes and annotation counts:")
for class_name, count in class_counts.most_common():
    print(f"{class_name}: {count}")

print("\nTotal annotated objects:", total_boxes)
print("XML annotations referencing missing images:", len(missing_images))

if missing_images:
    print("Examples of missing images:", missing_images[:10])

if xml_files:
    print("\nExample XML file:", xml_files[0].name)
    print(ET.tostring(
        ET.parse(xml_files[0]).getroot(),
        encoding="unicode"
    )[:2500])
```

##  Visualize and Validate Road-Damage Bounding Boxes

### Objective
The objective of this step is to visualize the road-damage annotations stored in the Pascal VOC XML files. Each annotation contains a damage class and bounding-box coordinates that identify the approximate location of the damage in an image.

### Methodology
1. Load a sample of training images and their corresponding XML annotation files.
2. Extract the image dimensions, damage class names, and bounding-box coordinates.
3. Draw rectangular bounding boxes around the annotated damage regions using Matplotlib.
4. Display the damage class label above each bounding box.
5. Inspect the images to verify whether the boxes correspond to the visible road damage.

### Libraries Used
- **OpenCV:** Reads road images.
- **xml.etree.ElementTree:** Extracts annotation information from XML files.
- **Matplotlib:** Displays images and bounding boxes.
- **pathlib:** Locates image and annotation files.

### Expected Outcome
The output will show sample road images with their annotated damage categories and bounding boxes. This visual inspection will help identify incorrect coordinates, poorly aligned boxes, unexpected labels, or other annotation issues before preparing the dataset for YOLO training.

### Importance
Object detection models learn from both the image pixels and the associated bounding-box labels. Incorrect annotations can confuse the model and reduce detection performance. Visual validation is therefore an essential quality-control step before converting the annotations into YOLO format.

```python

from pathlib import Path
import random
import xml.etree.ElementTree as ET

import cv2
import matplotlib.pyplot as plt
import matplotlib.patches as patches

dataset = Path("Road_Damage_Dataset")
image_dir = dataset / "images/train_images"
xml_dir = dataset / "labels/train_labels/xmls"

# Fixed seed makes the sample reproducible
random.seed(42)

xml_files = sorted(xml_dir.glob("*.xml"))
sample_files = random.sample(xml_files, min(6, len(xml_files)))

fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes = axes.flatten()

for ax, xml_file in zip(axes, sample_files):
    root = ET.parse(xml_file).getroot()
    filename = root.findtext("filename")
    image_path = image_dir / filename

    image = cv2.imread(str(image_path))

    if image is None:
        ax.set_title(f"Could not read: {filename}")
        ax.axis("off")
        continue

    # Convert BGR (OpenCV) to RGB (Matplotlib)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    height, width = image.shape[:2]

    ax.imshow(image)

    for obj in root.findall("object"):
        class_name = obj.findtext("name", default="Unknown")
        box = obj.find("bndbox")

        if box is None:
            continue

        xmin = float(box.findtext("xmin"))
        ymin = float(box.findtext("ymin"))
        xmax = float(box.findtext("xmax"))
        ymax = float(box.findtext("ymax"))

        # Check whether coordinates are valid
        if not (
            0 <= xmin < xmax <= width
            and 0 <= ymin < ymax <= height
        ):
            print("Invalid bounding box:", xml_file.name, class_name)
            continue

        rect = patches.Rectangle(
            (xmin, ymin),
            xmax - xmin,
            ymax - ymin,
            linewidth=2,
            edgecolor="red",
            facecolor="none"
        )
        ax.add_patch(rect)

        ax.text(
            xmin,
            max(0, ymin - 5),
            class_name,
            color="white",
            fontsize=9,
            bbox=dict(facecolor="red", alpha=0.8, pad=2)
        )

    ax.set_title(filename)
    ax.axis("off")

for ax in axes[len(sample_files):]:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

##  Phase 2: Dataset Preprocessing &amp; Taxonomy Standardization
---

### What Problem is happening ?
In multi-national datasets like **RDD2022**, raw annotation files often contain fine-grained sub-categories (e.g., `D01` for construction joint longitudinal cracks) or non-structural road markings (e.g., `D43` crosswalk blur, `D50` manholes). 

To train a robust global object detection model (e.g., YOLOv8 / Faster R-CNN), we consolidate these labels into **4 core structural damage classes** without discarding image samples or reducing dataset size.

---

### Dual Preprocessing Strategy

| Strategy | Applicable Classes | Action Taken | Rationale |
| :--- | :--- | :--- | :--- |
| **1. Class Mapping (Aggregation)** | `D01` → `D00`<br />`D11` → `D10` | Remap class IDs in annotation tags. | Retains 100% of crack bounding boxes by merging fine sub-types into primary structural orientation classes. |
| **2. Bounding Box Filtering** | `D43`, `D44`, `D50` | Drop specific bounding box nodes from XML/TXT. | Removes non-structural annotations while keeping the image as valid training background. |

---

### Formal Label Mapping Dictionary

$$
\text{Taxonomy\_Map} = \begin{cases} \text{'D00'}, \text{'D01'} \rightarrow \mathbf{0} & \text{(Longitudinal Crack)} \\ \text{'D10'}, \text{'D11'} \rightarrow \mathbf{1} & \text{(Transverse Crack)} \\ \text{'D20'} \rightarrow \mathbf{2} & \text{(Alligator Crack)} \\ \text{'D40'} \rightarrow \mathbf{3} & \text{(Pothole)} \\ \text{Others } (\text{D43, D44, D50}) \rightarrow \text{DROP} & \text{(Excluded Bounding Boxes)} \end{cases}
$$


---

###  Annotation Processing Logic &amp; Rules

When parsing raw PASCAL VOC (`.xml`) or YOLO (`.txt`) annotation files during pipeline execution, the parser follows these strict rules:

1. **Zero Data Loss Rule**: No image file is deleted during this step. Even if an image contains *only* excluded bounding boxes (e.g., `D43`), it is preserved as a negative sample (background) to reduce false positives.
2. **Selective Box Removal**: If an image contains both a valid target (e.g., `D40` Pothole) and an excluded label (e.g., `D50` Manhole), only the `D50` bounding box node is pruned from the markup.
3. **Coordinate Normalization**: Valid bounding box absolute coordinates \((x_{min}, y_{min}, x_{max}, y_{max})\) are mapped to normalized YOLO format values \((x_{center}, y_{center}, w, h) \in [0, 1]\) using image dimensions.
---


##  Create a Separate Folder for Processed Annotations

The original Pascal VOC XML annotations are stored in `labels/train_labels/xmls`.

We will create a separate folder named `labels/processed_xmls` to store the cleaned annotations.

The original XML files will remain unchanged. All label remapping, correction of `D0W0` to `D00`, and removal of unwanted bounding boxes (`D43`, `D44`, and `D50`) will be performed on copies.

All training images will be preserved, including images with empty annotations. The test images and their contents will remain untouched.


```python

from pathlib import Path
import shutil

# Dataset location
PROJECT_DIR = Path("Road_Damage_Dataset")

# Original XML annotations
SOURCE_XML_DIR = PROJECT_DIR / "labels" / "train_labels" / "xmls"

# Separate folder for cleaned XML annotations
PROCESSED_XML_DIR = PROJECT_DIR / "labels" / "processed_xmls"

# Verify the source folder
if not SOURCE_XML_DIR.is_dir():
    raise FileNotFoundError(
        f"Original XML folder not found: {SOURCE_XML_DIR}"
    )

# Safety check: do not overwrite an existing processed folder
if PROCESSED_XML_DIR.exists() and any(PROCESSED_XML_DIR.iterdir()):
    raise FileExistsError(
        f"{PROCESSED_XML_DIR} already contains files. "
        "Inspect it before copying; no files were changed."
    )

# Create destination folder
PROCESSED_XML_DIR.mkdir(parents=True, exist_ok=True)

# Copy XML annotations
xml_files = list(SOURCE_XML_DIR.glob("*.xml"))

for xml_file in xml_files:
    shutil.copy2(
        xml_file,
        PROCESSED_XML_DIR / xml_file.name
    )

print("XML copying completed.")
print("Original XML files:", len(xml_files))
print("Copied XML files:", len(list(PROCESSED_XML_DIR.glob("*.xml"))))
print("Original folder:", SOURCE_XML_DIR)
print("Processed folder:", PROCESSED_XML_DIR)
```

##  Clean and Standardize XML Annotations

We will preprocess the copied XML annotations using two operations:

**1. Class mapping**
- `D01` → `D00` (Longitudinal Crack)
- `D0w0` → `D00` (visually verified correction)
- `D11` → `D10` (Transverse Crack)

**2. Bounding-box filtering**
- Remove objects labeled `D43`, `D44`, and `D50`.
- Preserve all other valid target objects and their original bounding-box coordinates.

All images will be preserved, including images that contain only excluded objects. Such images will have empty XML annotations after cleaning.

The original XML files will not be modified. All operations will be applied to the copies in `labels/processed_xmls`.

```python
from pathlib import Path
import xml.etree.ElementTree as ET
import shutil
from collections import Counter

# Dataset paths
PROJECT_DIR = Path("Road_Damage_Dataset")

XML_DIR = PROJECT_DIR / "labels" / "processed_xmls"
BACKUP_DIR = PROJECT_DIR / "labels" / "xml_backup_before_cleaning"

# Label mapping and excluded classes
CLASS_MAPPING = {
    "D01": "D00",
    "D0w0": "D00",
    "D11": "D10",
}

EXCLUDED_CLASSES = {"D43", "D44", "D50"}

# Safety checks
if not XML_DIR.is_dir():
    raise FileNotFoundError(f"Processed XML folder not found: {XML_DIR}")

xml_files = sorted(XML_DIR.glob("*.xml"))

if len(xml_files) != 7706:
    raise ValueError(
        f"Expected 7706 XML files, but found {len(xml_files)}. "
        "Stop and verify the dataset before cleaning."
    )

if BACKUP_DIR.exists():
    raise FileExistsError(
        f"Backup folder already exists: {BACKUP_DIR}. "
        "Choose a new backup folder name before proceeding."
    )

# Back up the XML copies before modifying them
shutil.copytree(XML_DIR, BACKUP_DIR)

# Statistics
mapped_counts = Counter()
removed_counts = Counter()
final_counts = Counter()
empty_after_cleaning = 0
objects_before = 0
objects_after = 0

# Clean each copied XML
for xml_file in xml_files:
    tree = ET.parse(xml_file)
    root = tree.getroot()

    for obj in list(root.findall("object")):
        name_element = obj.find("name")

        if name_element is None or not name_element.text:
            continue

        original_name = name_element.text.strip()
        objects_before += 1

        # Remove excluded bounding boxes
        if original_name in EXCLUDED_CLASSES:
            root.remove(obj)
            removed_counts[original_name] += 1
            continue

        # Standardize class labels
        standardized_name = CLASS_MAPPING.get(
            original_name, original_name
        )

        if standardized_name != original_name:
            mapped_counts[original_name] += 1
            name_element.text = standardized_name

        final_counts[standardized_name] += 1
        objects_after += 1

    if not root.findall("object"):
        empty_after_cleaning += 1

    tree.write(
        xml_file,
        encoding="utf-8",
        xml_declaration=True
    )

print("XML cleaning completed successfully!")
print("\nObjects mapped:")
for name, count in sorted(mapped_counts.items()):
    print(f"{name} -> {CLASS_MAPPING[name]}: {count}")

print("\nBounding boxes removed:")
for name in sorted(EXCLUDED_CLASSES):
    print(f"{name}: {removed_counts[name]}")

print("\nFinal target class counts:")
for name in ["D00", "D10", "D20", "D40"]:
    print(f"{name}: {final_counts[name]}")

print("\nTotal objects before cleaning:", objects_before)
print("Total objects after cleaning:", objects_after)
print("XML files preserved:", len(xml_files))
print("Empty XML files after cleaning:", empty_after_cleaning)
print("Pre-cleaning backup:", BACKUP_DIR)
```

##  PHASE 3 :-Convert Cleaned XML Annotations to YOLO Format

The cleaned Pascal VOC XML annotations will be converted into YOLO-format text files.

Each object will be represented by five values:

Bounding-box coordinates will be normalized using the image width and height. The resulting center coordinates, width, and height should lie within the normalized range [0, 1].

###  Conversion of XML Annotations to YOLO Format

The cleaned Pascal VOC XML annotations were converted into YOLO-format text files. Each XML file was processed, and its bounding-box coordinates were converted into normalized YOLO coordinates.

#### 1. Class Mapping

The following class IDs were used:

| Class ID | Original Label | Class Name |
|---:|---|---|
| 0 | `D00` | Longitudinal |
| 1 | `D10` | Transverse |
| 2 | `D20` | Alligator |
| 3 | `D40` | Pothole |

#### 2. Bounding-Box Conversion

Each annotated object was represented using five values:

`class_id x_center y_center width height`

The center coordinates, width, and height were normalized using the image dimensions:

- `x_center` = (xmin + xmax) / (2 × image width)
- `y_center` = (ymin + ymax) / (2 × image height)
- `width` = (xmax − xmin) / image width
- `height` = (ymax − ymin) / image height

The conversion code also checked image dimensions, class labels, and bounding-box coordinates to identify invalid annotations.

#### 3. Conversion Results

| Metric | Expected Result |
|---|---:|
| XML files processed | 7,706 |
| YOLO TXT files created | 7,706 |
| Total objects converted | 7,056 |
| Empty YOLO label files | 4,387 |

*Note: These are the expected results and should be replaced with the actual output if the conversion produces different counts.*

#### 4. Handling Empty Annotations

A YOLO `.txt` file was created for every XML annotation file, including files with no retained target objects. These empty text files represent negative samples for the four selected classes.

#### 5. Output Location

The converted YOLO labels were saved in:

`labels/yolo_labels/`

The original XML annotations and their pre-cleaning backup were preserved.

#### Conclusion

The conversion process prepares the cleaned road-damage annotations for YOLO-based object detection. Before training, the generated labels must be validated to ensure that the class IDs, normalized coordinates, file counts, and image-label associations are correct.

```python
from pathlib import Path
import xml.etree.ElementTree as ET

# Main dataset folder
dataset_dir = Path("Road_Damage_Dataset")

# Input and output folders
xml_dir = dataset_dir / "labels" / "processed_xmls"
yolo_labels_dir = dataset_dir / "labels" / "yolo_labels"

# Map cleaned class names to YOLO class IDs
class_mapping = {
    "D00": 0,  # Longitudinal
    "D10": 1,  # Transverse
    "D20": 2,  # Alligator
    "D40": 3   # Pothole
}

converted_files = 0
total_objects = 0
empty_labels = 0

for xml_file in sorted(xml_dir.glob("*.xml")):
    tree = ET.parse(xml_file)
    root = tree.getroot()

    # Read image dimensions from XML
    size = root.find("size")

    if size is None:
        raise ValueError(f"Missing image size in {xml_file.name}")

    image_width = int(size.findtext("width", "0"))
    image_height = int(size.findtext("height", "0"))

    if image_width <= 0 or image_height <= 0:
        raise ValueError(f"Invalid image dimensions in {xml_file.name}")

    yolo_lines = []

    for obj in root.findall("object"):
        class_name = obj.findtext("name", "").strip()

        if class_name not in class_mapping:
            raise ValueError(
                f"Unexpected class '{class_name}' in {xml_file.name}"
            )

        bbox = obj.find("bndbox")

        if bbox is None:
            raise ValueError(f"Missing bounding box in {xml_file.name}")

        xmin = float(bbox.findtext("xmin", "nan"))
        ymin = float(bbox.findtext("ymin", "nan"))
        xmax = float(bbox.findtext("xmax", "nan"))
        ymax = float(bbox.findtext("ymax", "nan"))

        # Validate bounding-box coordinates
        if not (
            0 <= xmin < xmax <= image_width
            and 0 <= ymin < ymax <= image_height
        ):
            raise ValueError(
                f"Invalid bounding box in {xml_file.name}: "
                f"{xmin}, {ymin}, {xmax}, {ymax}"
            )

        # Convert Pascal VOC coordinates to YOLO format
        x_center = ((xmin + xmax) / 2) / image_width
        y_center = ((ymin + ymax) / 2) / image_height
        width = (xmax - xmin) / image_width
        height = (ymax - ymin) / image_height

        class_id = class_mapping[class_name]

        yolo_lines.append(
            f"{class_id} {x_center:.6f} {y_center:.6f} "
            f"{width:.6f} {height:.6f}"
        )

    # Create one TXT file for every XML, even if it is empty
    txt_file = yolo_labels_dir / f"{xml_file.stem}.txt"
    txt_file.write_text(
        "\n".join(yolo_lines) + ("\n" if yolo_lines else ""),
        encoding="utf-8"
    )

    converted_files += 1
    total_objects += len(yolo_lines)

    if not yolo_lines:
        empty_labels += 1

print("YOLO conversion completed successfully!")
print("XML files processed:", converted_files)
print("YOLO TXT files created:", len(list(yolo_labels_dir.glob("*.txt"))))
print("Total objects converted:", total_objects)
print("Empty YOLO label files:", empty_labels)
print("Output folder:", yolo_labels_dir)
```


# *Dataset Validation Overview*

> validating the structural integrity, bounding box coordinates, and class distributions of the Road Damage Dataset. By checking every line of the YOLO annotation files (.txt), the script ensures that the bounding boxes are mathematically sound, fully contained within image boundaries, and assigned to valid class IDs before model training.

```python
from pathlib import Path
from collections import Counter
import math

# Dataset path
dataset_dir = Path( "Road_Damage_Dataset")

yolo_labels_dir = dataset_dir / "labels" / "yolo_labels"

expected_classes = {0, 1, 2, 3}

label_files = list(yolo_labels_dir.glob("*.txt"))

class_counts = Counter()
total_objects = 0
empty_files = 0
invalid_lines = []
unexpected_classes = Counter()

for label_file in label_files:
    content = label_file.read_text(encoding="utf-8").strip()

    # Empty files are valid negative samples
    if not content:
        empty_files += 1
        continue

    for line_number, line in enumerate(content.splitlines(), start=1):
        parts = line.split()

        if len(parts) != 5:
            invalid_lines.append(
                (label_file.name, line_number, "Expected 5 values")
            )
            continue

        try:
            class_id = int(parts[0])
            x_center, y_center, width, height = map(float, parts[1:])
        except ValueError:
            invalid_lines.append(
                (label_file.name, line_number, "Non-numeric value")
            )
            continue

        if class_id not in expected_classes:
            unexpected_classes[class_id] += 1

        values = [x_center, y_center, width, height]

        if not all(math.isfinite(v) for v in values):
            invalid_lines.append(
                (label_file.name, line_number, "Non-finite coordinate")
            )
            continue

        if not all(0 <= v <= 1 for v in values):
            invalid_lines.append(
                (label_file.name, line_number, "Value outside [0, 1]")
            )
            continue

        if width <= 0 or height <= 0:
            invalid_lines.append(
                (label_file.name, line_number, "Non-positive box size")
            )
            continue

        # Check that the box remains within image boundaries
        if (
            x_center - width / 2 < -1e-6
            or x_center + width / 2 > 1 + 1e-6
            or y_center - height / 2 < -1e-6
            or y_center + height / 2 > 1 + 1e-6
        ):
            invalid_lines.append(
                (label_file.name, line_number, "Box exceeds image bounds")
            )
            continue

        class_counts[class_id] += 1
        total_objects += 1

print("YOLO label files:", len(label_files))
print("Empty label files:", empty_files)
print("Valid objects:", total_objects)

print("\nObjects per class:")
class_names = {
    0: "Longitudinal",
    1: "Transverse",
    2: "Alligator",
    3: "Pothole"
}

for class_id, class_name in class_names.items():
    print(f"{class_id} - {class_name}: {class_counts[class_id]}")

print("\nUnexpected class IDs:", unexpected_classes or "None")
print("Invalid annotation lines:", len(invalid_lines))

if invalid_lines:
    print("\nFirst 10 invalid entries:")
    for item in invalid_lines[:10]:
        print(item)
```

```python
from pathlib import Path
import random
import xml.etree.ElementTree as ET

import cv2
import matplotlib.pyplot as plt
import matplotlib.patches as patches

# --------------------------------------------------
# 1. Define dataset paths
# --------------------------------------------------

project_path = Path("")

image_dir = project_path / "Road_Damage_YOLO_Split" / "images" / "train"
yolo_label_dir = project_path / "Road_Damage_YOLO_Split" / "labels" / "train"

# Processed XML annotations from the original preprocessing stage
xml_dir = project_path / "Road_Damage_Dataset" / "labels" / "processed_xmls"

# --------------------------------------------------
# 2. Define class names and colors
# --------------------------------------------------

class_names = {
    "D00": "Longitudinal",
    "D10": "Transverse",
    "D20": "Alligator",
    "D40": "Pothole"
}

yolo_class_names = {
    0: "Longitudinal",
    1: "Transverse",
    2: "Alligator",
    3: "Pothole"
}

# --------------------------------------------------
# 3. Find images with both XML and YOLO annotations
# --------------------------------------------------

image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

if not image_dir.is_dir():
    raise FileNotFoundError(f"Image directory not found: {image_dir}")

if not yolo_label_dir.is_dir():
    raise FileNotFoundError(f"YOLO label directory not found: {yolo_label_dir}")

if not xml_dir.is_dir():
    raise FileNotFoundError(f"Processed XML directory not found: {xml_dir}")

image_paths = []

for image_path in image_dir.iterdir():
    if image_path.suffix.lower() not in image_extensions:
        continue

    xml_path = xml_dir / f"{image_path.stem}.xml"
    yolo_path = yolo_label_dir / f"{image_path.stem}.txt"

    if xml_path.exists() and yolo_path.exists():
        image_paths.append(image_path)

if len(image_paths) < 6:
    raise ValueError(
        f"Only {len(image_paths)} images have both XML and YOLO files. "
        "Please check the filenames and paths."
    )

# --------------------------------------------------
# 4. Select six random examples
# --------------------------------------------------

random.seed(42)
sample_images = random.sample(image_paths, 6)

# --------------------------------------------------
# 5. Helper function to read XML bounding boxes
# --------------------------------------------------

def read_xml_boxes(xml_path):
    root = ET.parse(xml_path).getroot()
    boxes = []

    for obj in root.findall("object"):
        class_id = obj.findtext("name", default="Unknown").strip()
        box = obj.find("bndbox")

        if box is None:
            continue

        xmin = float(box.findtext("xmin"))
        ymin = float(box.findtext("ymin"))
        xmax = float(box.findtext("xmax"))
        ymax = float(box.findtext("ymax"))

        # XML boxes use absolute pixel coordinates.
        # Clamp to image boundaries later when drawing.
        boxes.append((class_id, xmin, ymin, xmax, ymax))

    return boxes

# --------------------------------------------------
# 6. Helper function to read YOLO bounding boxes
# --------------------------------------------------

def read_yolo_boxes(label_path, image_width, image_height):
    boxes = []

    with open(label_path, "r", encoding="utf-8") as file:
        for line_number, line in enumerate(file, start=1):
            values = line.strip().split()

            if not values:
                continue

            if len(values) != 5:
                print(
                    f"Invalid YOLO line in {label_path.name}, "
                    f"line {line_number}: {line.strip()}"
                )
                continue

            class_id = int(values[0])
            x_center, y_center, box_width, box_height = map(
                float, values[1:]
            )

            # Convert normalized YOLO coordinates to pixel coordinates.
            xmin = (x_center - box_width / 2) * image_width
            ymin = (y_center - box_height / 2) * image_height
            xmax = (x_center + box_width / 2) * image_width
            ymax = (y_center + box_height / 2) * image_height

            boxes.append(
                (class_id, xmin, ymin, xmax, ymax)
            )

    return boxes

# --------------------------------------------------
# 7. Draw the XML and YOLO boxes side by side
# --------------------------------------------------

fig, axes = plt.subplots(6, 2, figsize=(16, 30))

for row, image_path in enumerate(sample_images):
    xml_path = xml_dir / f"{image_path.stem}.xml"
    yolo_path = yolo_label_dir / f"{image_path.stem}.txt"

    image_bgr = cv2.imread(str(image_path))

    if image_bgr is None:
        print(f"Could not read image: {image_path}")
        axes[row, 0].axis("off")
        axes[row, 1].axis("off")
        continue

    image = cv2.cvtColor(image_bgr, cv2.COLOR_BGR2RGB)
    height, width = image.shape[:2]

    xml_boxes = read_xml_boxes(xml_path)
    yolo_boxes = read_yolo_boxes(yolo_path, width, height)

    # Left: original processed XML annotations
    axes[row, 0].imshow(image)

    for class_id, xmin, ymin, xmax, ymax in xml_boxes:
        xmin = max(0, min(width - 1, xmin))
        ymin = max(0, min(height - 1, ymin))
        xmax = max(0, min(width - 1, xmax))
        ymax = max(0, min(height - 1, ymax))

        rect = patches.Rectangle(
            (xmin, ymin),
            xmax - xmin,
            ymax - ymin,
            linewidth=2,
            edgecolor="red",
            facecolor="none"
        )
        axes[row, 0].add_patch(rect)

        label = class_names.get(class_id, class_id)
        axes[row, 0].text(
            xmin,
            max(ymin - 4, 0),
            label,
            color="white",
            fontsize=8,
            bbox=dict(facecolor="red", alpha=0.8, pad=2)
        )

    axes[row, 0].set_title(
        f"Processed XML — {image_path.name}\n"
        f"Boxes: {len(xml_boxes)}"
    )
    axes[row, 0].axis("off")

    # Right: converted YOLO TXT annotations
    axes[row, 1].imshow(image)

    for class_id, xmin, ymin, xmax, ymax in yolo_boxes:
        xmin = max(0, min(width - 1, xmin))
        ymin = max(0, min(height - 1, ymin))
        xmax = max(0, min(width - 1, xmax))
        ymax = max(0, min(height - 1, ymax))

        rect = patches.Rectangle(
            (xmin, ymin),
            xmax - xmin,
            ymax - ymin,
            linewidth=2,
            edgecolor="lime",
            facecolor="none"
        )
        axes[row, 1].add_patch(rect)

        label = yolo_class_names.get(class_id, f"Unknown ID {class_id}")
        axes[row, 1].text(
            xmin,
            max(ymin - 4, 0),
            label,
            color="black",
            fontsize=8,
            bbox=dict(facecolor="lime", alpha=0.8, pad=2)
        )

    axes[row, 1].set_title(
        f"YOLO TXT — {image_path.name}\n"
        f"Boxes: {len(yolo_boxes)}"
    )
    axes[row, 1].axis("off")

plt.suptitle(
    "Annotation Comparison: Processed XML (Left) vs YOLO TXT (Right)",
    fontsize=16,
    y=1.002
)
plt.tight_layout()
plt.show()

print("Displayed six random examples.")
print("Left: processed XML annotations")
print("Right: converted YOLO TXT annotations")
print("Compare box positions, sizes, class names, and box counts.")
```

##  Create Fixed Training, Validation, and Test Splits

The 7,706 labelled images will be divided into three subsets:

- Training set: 75% — used to train the baseline YOLO model.
- Validation set: 10% — used for model selection and hyperparameter tuning.
- Fixed test set: 15% — reserved for final quantitative evaluation.

The same training, validation, and test split will be used for both experiments:
1. Model A: trained on real training data only.
2. Model B: trained on the same real training data plus synthetic data.

The fixed test set will not be used for training, hyperparameter tuning, synthetic-data generation, or pseudo-labelling.

The approximately 2,000 unlabelled test images will remain separate and will not be included in these splits.

The split will use a fixed random seed for reproducibility. Original images and YOLO labels will not be moved, modified, or deleted.

```python
from pathlib import Path
import random
import csv

# Dataset paths
project_dir = Path("Road_Damage_Dataset")

image_dir = project_dir / "images" / "train_images"
label_dir = project_dir / "labels" / "yolo_labels"

# Save the split manifest separately from the original dataset
output_dir = project_dir / "split_manifests"
output_dir.mkdir(parents=True, exist_ok=True)

manifest_path = output_dir / "real_data_split.csv"

# Supported image formats
image_extensions = {".jpg", ".jpeg", ".png", ".bmp"}

# Find all training images
images = sorted(
    p for p in image_dir.iterdir()
    if p.is_file() and p.suffix.lower() in image_extensions
)

# Check that every image has a corresponding label file
missing_labels = [
    image.name
    for image in images
    if not (label_dir / f"{image.stem}.txt").exists()
]

# Check for labels without matching images
image_stems = {image.stem for image in images}

orphan_labels = [
    label.name
    for label in label_dir.glob("*.txt")
    if label.stem not in image_stems
]

print("Training images found:", len(images))
print("Images without label files:", len(missing_labels))
print("Label files without matching images:", len(orphan_labels))

if missing_labels or orphan_labels:
    raise ValueError(
        "Image-label matching failed. Fix the mismatches before splitting."
    )

if len(images) != 7706:
    raise ValueError(
        f"Expected 7706 images, but found {len(images)}. "
        "Check the dataset paths and image count before continuing."
    )

# Reproducible shuffle
random_seed = 42
random_generator = random.Random(random_seed)
random_generator.shuffle(images)

# Calculate split sizes
total = len(images)
train_count = int(total * 0.75)
val_count = int(total * 0.10)
test_count = total - train_count - val_count

train_images = images[:train_count]
val_images = images[train_count:train_count + val_count]
test_images = images[train_count + val_count:]

# Save image and matching label filenames
with manifest_path.open("w", newline="", encoding="utf-8") as file:
    writer = csv.writer(file)
    writer.writerow(["image", "label", "split"])

    for split_name, split_images in [
        ("train", train_images),
        ("val", val_images),
        ("test", test_images),
    ]:
        for image in split_images:
            writer.writerow([
                image.name,
                f"{image.stem}.txt",
                split_name,
            ])

print("\nSplit counts:")
print("Training:", len(train_images))
print("Validation:", len(val_images))
print("Fixed test:", len(test_images))
print("Total:", len(train_images) + len(val_images) + len(test_images))
print("\nManifest saved to:", manifest_path)
```

##  Organize the Dataset for YOLO Training

The split manifest created above will be used to organize the existing images and YOLO labels into separate training, validation, and test folders.

A new dataset directory will be created with the following structure:

- `images/train` and `labels/train`
- `images/val` and `labels/val`
- `images/test` and `labels/test`

Each image will be copied together with its corresponding `.txt` label file. Empty label files will also be preserved.

The original images, XML annotations, and YOLO label files will remain unchanged. The separate 2,000 unlabelled images will not be included in this dataset.

```python
from pathlib import Path
import csv
import shutil

# Original dataset
project_dir = Path( "Road_Damage_Dataset")

image_dir = project_dir / "images" / "train_images"
label_dir = project_dir / "labels" / "yolo_labels"
manifest_path = project_dir / "split_manifests" / "real_data_split.csv"

# Create the new dataset OUTSIDE the original project folder
destination = Path("Road_Damage_YOLO_Split")

# Safety checks
if not manifest_path.exists():
    raise FileNotFoundError(f"Split manifest not found: {manifest_path}")

if destination.exists():
    raise FileExistsError(
        f"Destination already exists: {destination}\n"
        "Choose a new destination folder or inspect the existing folder first."
    )

# Read the split manifest
with manifest_path.open("r", newline="", encoding="utf-8") as file:
    rows = list(csv.DictReader(file))

if len(rows) != 7706:
    raise ValueError(
        f"Expected 7706 manifest rows, but found {len(rows)}."
    )

# Validate all source files before copying anything
for row in rows:
    image_path = image_dir / row["image"]
    label_path = label_dir / row["label"]

    if not image_path.is_file():
        raise FileNotFoundError(f"Image not found: {image_path}")

    if not label_path.is_file():
        raise FileNotFoundError(f"Label not found: {label_path}")

    if row["split"] not in {"train", "val", "test"}:
        raise ValueError(f"Unexpected split: {row['split']}")

# Create the standard YOLO folder structure
for split in ("train", "val", "test"):
    (destination / "images" / split).mkdir(parents=True)
    (destination / "labels" / split).mkdir(parents=True)

# Copy each image and its matching label
counts = {"train": 0, "val": 0, "test": 0}

for row in rows:
    split = row["split"]

    source_image = image_dir / row["image"]
    source_label = label_dir / row["label"]

    target_image = destination / "images" / split / row["image"]
    target_label = destination / "labels" / split / row["label"]

    shutil.copy2(source_image, target_image)
    shutil.copy2(source_label, target_label)

    counts[split] += 1

# Verify the copied files
print("Dataset created successfully!")
print("Destination:", destination)

for split in ("train", "val", "test"):
    image_count = sum(
        1 for p in (destination / "images" / split).iterdir()
        if p.is_file()
    )
    label_count = sum(
        1 for p in (destination / "labels" / split).glob("*.txt")
        if p.is_file()
    )

    print(
        f"{split}: {image_count} images, "
        f"{label_count} label files"
    )

    if image_count != counts[split] or label_count != counts[split]:
        raise RuntimeError(f"File count verification failed for {split}.")

print("Total images copied:", sum(counts.values()))
print("Original dataset remains unchanged.")
```

## Handling mismatch 

The Computer Vision Dataset and LLM Dataset were processed separately. Each image filename and label filename was matched using its final numeric identifier. Successfully matched label files were copied into a new `train_labels_matched` folder and renamed to correspond to their images, while the original labels remained unchanged. Unmatched or ambiguous images were reported for review to help ensure correct image–annotation pairing before YOLO training.

---


```python
from pathlib import Path
import re
import shutil

# Project directory on your Mac
project_dir = Path(
    "/Users/anubhav/Downloads/ML-Project-LossLess-/MAIN_FOLDER"
)

synthetic_root = project_dir / "synthetic_data_for _YOLO"

sources = [
    synthetic_root / "Computer_Vision_Dataset",
    synthetic_root / "LLM_Dataset",
]

image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}


def get_numeric_id(filename):
    """Extract the final number from a filename."""
    match = re.search(r"(\d+)$", Path(filename).stem)
    return int(match.group(1)) if match else None


for source in sources:
    image_dir = source / "train_images"
    label_dir = source / "train_labels"

    if not image_dir.is_dir() or not label_dir.is_dir():
        print(f"\nSkipping missing folders: {source}")
        continue

    # Create a new folder; original labels remain untouched.
    fixed_label_dir = source / "train_labels_matched"
    fixed_label_dir.mkdir(exist_ok=False)

    images = sorted(
        p for p in image_dir.iterdir()
        if p.is_file() and p.suffix.lower() in image_extensions
    )

    labels = sorted(label_dir.glob("*.txt"))

    # Group labels by numeric ID.
    labels_by_id = {}
    for label in labels:
        number = get_numeric_id(label.name)
        if number is not None:
            labels_by_id.setdefault(number, []).append(label)

    matched = 0
    unmatched_images = []
    ambiguous_images = []

    for image in images:
        number = get_numeric_id(image.name)

        if number is None:
            unmatched_images.append(image.name)
            continue

        candidates = labels_by_id.get(number, [])

        if len(candidates) == 0:
            unmatched_images.append(image.name)
            continue

        if len(candidates) > 1:
            ambiguous_images.append(image.name)
            continue

        label = candidates[0]

        # Copy the label using the corresponding image's stem.
        destination = fixed_label_dir / f"{image.stem}.txt"
        shutil.copy2(label, destination)
        matched += 1

    print(f"\nDataset: {source.name}")
    print("Images:", len(images))
    print("Original label files:", len(labels))
    print("Successfully matched:", matched)
    print("Images without a numeric-ID match:", len(unmatched_images))
    print("Images with ambiguous matches:", len(ambiguous_images))
    print("Matched labels saved to:", fixed_label_dir)

    if unmatched_images:
        print("Unmatched examples:", unmatched_images[:10])

    if ambiguous_images:
        print("Ambiguous examples:", ambiguous_images[:10])

    if matched == len(images):
        print("All images have a numeric-ID match.")
    else:
        print("Review unmatched images before using this dataset.")
```

In this step, we will create a separate YOLO dataset configuration for Model B. The training split will contain real training images and correctly matched synthetic images with their annotations.

The validation and fixed test splits will contain only real images from the existing Model A dataset. Synthetic images will not be included in either split.

The original dataset and Model A's `data.yaml` will remain unchanged. A separate configuration file, `data_real_synthetic.yaml`, will be created for Model B with the same four classes: Longitudinal, Transverse, Alligator, and Pothole.

---

```python
from pathlib import Path
import shutil
import yaml

# --------------------------------------------------
# 1. Project paths
# --------------------------------------------------

project_dir = Path(
    "/Users/anubhav/Downloads/ML-Project-LossLess-/MAIN_FOLDER"
)

# Model A dataset; its data.yaml will not be changed.
real_dir = project_dir / "Road_Damage_YOLO_Split"

synthetic_root = project_dir / "synthetic_data_for _YOLO"

synthetic_sources = [
    synthetic_root / "Computer_Vision_Dataset",
    synthetic_root / "LLM_Dataset",
]

# Separate dataset for Model B
model_b_dir = project_dir / "Road_Damage_YOLO_Real_Synthetic"

image_extensions = {".jpg", ".jpeg", ".png", ".bmp", ".webp"}

class_names = {
    0: "Longitudinal",
    1: "Transverse",
    2: "Alligator",
    3: "Pothole",
}

# --------------------------------------------------
# 2. Validate existing Model A folders
# --------------------------------------------------

for split in ["train", "val", "test"]:
    for kind in ["images", "labels"]:
        folder = real_dir / kind / split
        if not folder.is_dir():
            raise FileNotFoundError(f"Missing Model A folder: {folder}")

# Model A configuration must already exist.
model_a_yaml = real_dir / "data.yaml"

if not model_a_yaml.is_file():
    raise FileNotFoundError(f"Model A YAML not found: {model_a_yaml}")

# Do not overwrite a previous Model B dataset.
if model_b_dir.exists():
    raise FileExistsError(
        f"Model B folder already exists: {model_b_dir}\n"
        "Rename or inspect it before running this cell again."
    )

# --------------------------------------------------
# 3. Create separate split folders
# --------------------------------------------------

for split in ["train", "val", "test"]:
    (model_b_dir / "images" / split).mkdir(
        parents=True, exist_ok=True
    )
    (model_b_dir / "labels" / split).mkdir(
        parents=True, exist_ok=True
    )

def copy_split(source_images, source_labels, dest_images, dest_labels):
    """Copy images and their matching YOLO labels."""
    images = sorted(
        p for p in source_images.iterdir()
        if p.is_file() and p.suffix.lower() in image_extensions
    )

    image_stems = {p.stem for p in images}
    label_stems = {p.stem for p in source_labels.glob("*.txt")}

    if image_stems != label_stems:
        raise ValueError(
            f"Image/label mismatch in {source_images}\n"
            f"Images without labels: {sorted(image_stems - label_stems)[:10]}\n"
            f"Labels without images: {sorted(label_stems - image_stems)[:10]}"
        )

    for image in images:
        shutil.copy2(image, dest_images / image.name)
        shutil.copy2(
            source_labels / f"{image.stem}.txt",
            dest_labels / f"{image.stem}.txt"
        )

    return len(images)

# --------------------------------------------------
# 4. Copy real training images and labels
# --------------------------------------------------

real_train_count = copy_split(
    real_dir / "images" / "train",
    real_dir / "labels" / "train",
    model_b_dir / "images" / "train",
    model_b_dir / "labels" / "train",
)

# --------------------------------------------------
# 5. Add synthetic training images and matched labels
# --------------------------------------------------

synthetic_counts = {}

for source_index, source in enumerate(synthetic_sources, start=1):
    image_dir = source / "train_images"
    matched_label_dir = source / "train_labels_matched"

    if not image_dir.is_dir():
        raise FileNotFoundError(f"Missing synthetic images: {image_dir}")

    if not matched_label_dir.is_dir():
        raise FileNotFoundError(
            f"Missing matched labels: {matched_label_dir}\n"
            "Run the synthetic matching step first."
        )

    count = copy_split(
        image_dir,
        matched_label_dir,
        model_b_dir / "images" / "train",
        model_b_dir / "labels" / "train",
    ) if False else None

    # Copy with a unique prefix to prevent filename collisions.
    images = sorted(
        p for p in image_dir.iterdir()
        if p.is_file() and p.suffix.lower() in image_extensions
    )

    image_stems = {p.stem for p in images}
    label_stems = {p.stem for p in matched_label_dir.glob("*.txt")}

    if image_stems != label_stems:
        raise ValueError(
            f"Matched-label mismatch in {source.name}.\n"
            f"Missing labels: {sorted(image_stems - label_stems)[:10]}\n"
            f"Extra labels: {sorted(label_stems - image_stems)[:10]}"
        )

    for image in images:
        new_stem = f"synthetic_{source_index}_{image.stem}"

        shutil.copy2(
            image,
            model_b_dir / "images" / "train" /
            f"{new_stem}{image.suffix.lower()}"
        )
        shutil.copy2(
            matched_label_dir / f"{image.stem}.txt",
            model_b_dir / "labels" / "train" / f"{new_stem}.txt"
        )

    synthetic_counts[source.name] = len(images)

# --------------------------------------------------
# 6. Copy ONLY real validation and test data
# --------------------------------------------------

real_val_count = copy_split(
    real_dir / "images" / "val",
    real_dir / "labels" / "val",
    model_b_dir / "images" / "val",
    model_b_dir / "labels" / "val",
)

real_test_count = copy_split(
    real_dir / "images" / "test",
    real_dir / "labels" / "test",
    model_b_dir / "images" / "test",
    model_b_dir / "labels" / "test",
)

# --------------------------------------------------
# 7. Create Model B YAML with a different filename
# --------------------------------------------------

data_config = {
    "path": str(model_b_dir.resolve()),
    "train": "images/train",
    "val": "images/val",
    "test": "images/test",
    "names": class_names,
}

yaml_path = model_b_dir / "data_real_synthetic.yaml"

with yaml_path.open("w", encoding="utf-8") as file:
    yaml.safe_dump(data_config, file, sort_keys=False)

# --------------------------------------------------
# 8. Print summary
# --------------------------------------------------

print("Model B dataset created successfully!")
print("Model A YAML preserved:", model_a_yaml)
print("Model B YAML:", yaml_path)
print("\nReal training images:", real_train_count)

for source_name, count in synthetic_counts.items():
    print(f"Synthetic training images ({source_name}):", count)

print(
    "Total training images:",
    real_train_count + sum(synthetic_counts.values())
)
print("Real validation images:", real_val_count)
print("Real fixed-test images:", real_test_count)
print("\nSynthetic images are included in training only.")
```

We evaluate both trained YOLO11n checkpoints on the same fixed real-world test set. We compare precision, recall, mAP@50, and mAP@50–95 to assess whether adding synthetic training data improves detection performance on real road images.

## Conclusion: Real-Only vs. Real + Synthetic Model Evaluation

The two YOLO11n models were evaluated on the same fixed real-world test dataset to investigate whether adding synthetic road-damage images to the training data improves detection performance.

The real-only baseline achieved a precision of 29.31%, recall of 32.92%, mAP@50 of 30.08%, and mAP@50–95 of 12.75%. The real-plus-synthetic model achieved a precision of 59.80%, recall of 32.97%, mAP@50 of 31.34%, and mAP@50–95 of 13.41%.

The results show that the real-plus-synthetic model substantially improved precision, while recall remained almost unchanged. The improvements in mAP@50 and mAP@50–95 were relatively small, increasing by 1.26 and 0.66 percentage points, respectively. This suggests that synthetic training data may help reduce incorrect detections, but it did not substantially increase the proportion of actual road damage detected.

Overall, the results provide preliminary evidence that incorporating synthetic data can improve certain aspects of road-damage detection on real images. However, the observed differences do not establish statistical significance. Further statistical analysis and repeated experiments are needed before concluding that synthetic data produces a reliable improvement in overall detection performance.

**Hypothesis decision:** The results are directionally consistent with the Null hypothesis that adding synthetic data improves detection performance. However, the Alternative hypothesis cannot yet be rejected without appropriate statistical evidence.
