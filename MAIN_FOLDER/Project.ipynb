## 2. Problem Statement

Road damage, including potholes and cracks, can cause accidents, vehicle damage, traffic congestion, and increased road maintenance costs.

Manual road inspection requires time, labour, and regular monitoring. Although machine learning models can automatically detect road damage, their performance may decrease under unfamiliar road conditions, poor lighting, rainfall, and differences in road surfaces.

Furthermore, publicly available datasets may not contain enough representative examples of rare damage types or difficult environmental conditions.

**Our problem is to investigate whether generating synthetic road-damage images and combining them with real road images can improve the accuracy and generalization of machine learning models used for road damage detection in Indian conditions.**

## 7. Synthetic Data Generation

Synthetic data is artificially generated data designed to represent real-world examples.

The project will investigate whether synthetic images can improve road-damage detection when real training examples are limited.

### Method 1: Image Compositing

- Select a real road image as the background.
- Extract or create a pothole or crack patch.
- Place the patch in a suitable road region.
- Blend it using OpenCV techniques such as Poisson blending.
- Generate the corresponding bounding-box annotation.

### Method 2: GAN-Based Generation

- Investigate a Generative Adversarial Network for generating road-damage appearances.
- Generate additional samples with different damage shapes and textures.
- Check image quality and label accuracy before using them for training.

### Method 3: Diffusion-Based Generation

- Explore a pretrained image-generation model with inpainting capabilities.
- Generate damage under varied road textures, lighting and weather conditions.
- Verify that generated defects look plausible and that annotations accurately identify the damage.

### Planned Starting Approach

Begin with image compositing because it is relatively simple and computationally inexpensive. Explore GANs or diffusion models if time and computing resources permit.

Synthetic images must be generated only for the training partition. The validation and test sets must remain independent real-world datasets.

## 11. Evaluation Metrics

- **Precision:** Proportion of predicted detections that are correct.
- **Recall:** Proportion of actual road-damage instances detected.
- **F1-score:** Balance between precision and recall.
- **mAP@50:** Mean average precision using an intersection-over-union threshold of 0.50.
- **mAP@50:95:** Mean average precision averaged over multiple intersection-over-union thresholds.
- **Per-class recall:** Performance for individual damage categories.
- **Inference time:** Time required to process an image.

The primary focus will be on mAP, recall and generalization to unseen real-world road images.

## Step 3: Inspect XML Annotations and Identify Road-Damage Classes

### Objective
The objective of this step is to examine the XML annotation files associated with the road-damage dataset. These annotations contain information about the image filenames, damage categories, and bounding-box coordinates required for training an object detection model such as YOLO.

### Methodology
1. **Locate the dataset:** Define the paths to the road-damage dataset, training images, and XML annotation directory using Python's `pathlib` module.
2. **Count annotation files:** Identify the number of XML files available for the training dataset.
3. **Extract damage categories:** Parse each XML file using `xml.etree.ElementTree` and count the annotated objects belonging to each damage class.
4. **Verify image availability:** Check whether the image filenames recorded in the XML annotations correspond to files present in the training image directory.
5. **Count annotated objects:** Calculate the total number of road-damage instances represented by the bounding-box annotations.
6. **Inspect a sample annotation:** Display part of an XML file to understand its structure and identify the information needed for later conversion into YOLO format.

### Libraries Used
- **Pathlib:** Locates dataset folders and files.
- **xml.etree.ElementTree:** Reads and parses XML annotation files.
- **collections.Counter:** Counts the occurrences of each road-damage category.

### Expected Outcome
This inspection will help us understand the dataset's annotation structure

```python

from pathlib import Path
import xml.etree.ElementTree as ET
from collections import Counter

dataset = Path("Road_Damage_Dataset")

xml_dir = dataset / "labels/train_labels/xmls"
train_dir = dataset / "images/train_images"

xml_files = list(xml_dir.glob("*.xml"))

print("Total XML files:", len(xml_files))
print("Training image folder exists:", train_dir.exists())

class_counts = Counter()
missing_images = []
total_boxes = 0

for xml_file in xml_files:
    try:
        tree = ET.parse(xml_file)
        root = tree.getroot()

        filename = root.findtext("filename")

        if filename:
            image_path = train_dir / filename
            if not image_path.exists():
                missing_images.append(filename)

        for obj in root.findall("object"):
            name = obj.findtext("name", default="Unknown").strip()
            class_counts[name] += 1
            total_boxes += 1

    except ET.ParseError:
        print("Invalid XML:", xml_file.name)

print("\nDamage classes and annotation counts:")
for class_name, count in class_counts.most_common():
    print(f"{class_name}: {count}")

print("\nTotal annotated objects:", total_boxes)
print("XML annotations referencing missing images:", len(missing_images))

if missing_images:
    print("Examples of missing images:", missing_images[:10])

if xml_files:
    print("\nExample XML file:", xml_files[0].name)
    print(ET.tostring(
        ET.parse(xml_files[0]).getroot(),
        encoding="unicode"
    )[:2500])
```

## Step 4: Visualize and Validate Road-Damage Bounding Boxes

### Objective
The objective of this step is to visualize the road-damage annotations stored in the Pascal VOC XML files. Each annotation contains a damage class and bounding-box coordinates that identify the approximate location of the damage in an image.

### Methodology
1. Load a sample of training images and their corresponding XML annotation files.
2. Extract the image dimensions, damage class names, and bounding-box coordinates.
3. Draw rectangular bounding boxes around the annotated damage regions using Matplotlib.
4. Display the damage class label above each bounding box.
5. Inspect the images to verify whether the boxes correspond to the visible road damage.

### Libraries Used
- **OpenCV:** Reads road images.
- **xml.etree.ElementTree:** Extracts annotation information from XML files.
- **Matplotlib:** Displays images and bounding boxes.
- **pathlib:** Locates image and annotation files.

### Expected Outcome
The output will show sample road images with their annotated damage categories and bounding boxes. This visual inspection will help identify incorrect coordinates, poorly aligned boxes, unexpected labels, or other annotation issues before preparing the dataset for YOLO training.

### Importance
Object detection models learn from both the image pixels and the associated bounding-box labels. Incorrect annotations can confuse the model and reduce detection performance. Visual validation is therefore an essential quality-control step before converting the annotations into YOLO format.

```python

from pathlib import Path
import random
import xml.etree.ElementTree as ET

import cv2
import matplotlib.pyplot as plt
import matplotlib.patches as patches

dataset = Path("MAIN_FOLDER/Road_Damage_Dataset")
image_dir = dataset / "images/train_images"
xml_dir = dataset / "labels/train_labels/xmls"

# Fixed seed makes the sample reproducible
random.seed(42)

xml_files = sorted(xml_dir.glob("*.xml"))
sample_files = random.sample(xml_files, min(6, len(xml_files)))

fig, axes = plt.subplots(2, 3, figsize=(16, 10))
axes = axes.flatten()

for ax, xml_file in zip(axes, sample_files):
    root = ET.parse(xml_file).getroot()
    filename = root.findtext("filename")
    image_path = image_dir / filename

    image = cv2.imread(str(image_path))

    if image is None:
        ax.set_title(f"Could not read: {filename}")
        ax.axis("off")
        continue

    # Convert BGR (OpenCV) to RGB (Matplotlib)
    image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    height, width = image.shape[:2]

    ax.imshow(image)

    for obj in root.findall("object"):
        class_name = obj.findtext("name", default="Unknown")
        box = obj.find("bndbox")

        if box is None:
            continue

        xmin = float(box.findtext("xmin"))
        ymin = float(box.findtext("ymin"))
        xmax = float(box.findtext("xmax"))
        ymax = float(box.findtext("ymax"))

        # Check whether coordinates are valid
        if not (
            0 <= xmin < xmax <= width
            and 0 <= ymin < ymax <= height
        ):
            print("Invalid bounding box:", xml_file.name, class_name)
            continue

        rect = patches.Rectangle(
            (xmin, ymin),
            xmax - xmin,
            ymax - ymin,
            linewidth=2,
            edgecolor="red",
            facecolor="none"
        )
        ax.add_patch(rect)

        ax.text(
            xmin,
            max(0, ymin - 5),
            class_name,
            color="white",
            fontsize=9,
            bbox=dict(facecolor="red", alpha=0.8, pad=2)
        )

    ax.set_title(filename)
    ax.axis("off")

for ax in axes[len(sample_files):]:
    ax.axis("off")

plt.tight_layout()
plt.show()
```

## Step 5: Investigate Unexpected Road-Damage Labels

### Objective
Identify the meaning and visual characteristics of additional class labels found in the XML annotations, including D44, D01, D43, D11, D50, and D0w0.

### Methodology
- Find XML files containing each unexpected class.
- Record the image filenames and bounding-box coordinates.
- Display representative examples for visual inspection.
- Compare the observed damage with the project's four selected categories.

### Expected Outcome
Determine whether the unexpected labels represent valid additional damage categories, inconsistent annotation codes, or possible data-entry errors. Use this evidence to decide which annotations to retain for the four-class detection task.


```python
from pathlib import Path
import xml.etree.ElementTree as ET
import cv2
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import random

# Use .resolve() to ensure paths handle relative layouts correctly
dataset = Path("MAIN_FOLDER/Road_Damage_Dataset").resolve()
image_dir = dataset / "images/train_images"
xml_dir = dataset / "labels/train_labels/xmls"

target_classes = ["D44", "D01", "D43", "D11", "D50", "D0w0"]

# Find XML files containing each target class
examples = {name: [] for name in target_classes}

# Check if directory exists first
if not xml_dir.exists():
    print(f"Error: XML directory not found at {xml_dir}")
else:
    for xml_file in xml_dir.glob("*.xml"):
        try:
            root = ET.parse(xml_file).getroot()
            for obj in root.findall("object"):
                name = obj.findtext("name", default="Unknown").strip()
                # Use case-insensitive checking if needed, or stick to exact match
                if name in examples and len(examples[name]) < 3:
                    examples[name].append(xml_file)
        except ET.ParseError:
            continue
            
print("Representative XML files for each class:")
for name, files in examples.items():
    print(f"\n{name}: Found {len(files)} files")
    for file in files:
        print("  ", file.name)

# Visualize one example for each class
random.seed(42)
selected = [
    (name, random.choice(files))
    for name, files in examples.items()
    if files  # This safely filters out classes with 0 found files
]

# CRITICAL FIX: If no images match, don't run the plotting logic
if not selected:
    print("\n[!] No matching classes found in your dataset. Skipping visualization pipeline.")
else:
    fig, axes = plt.subplots(
        len(selected), 1,
        figsize=(10, 5 * len(selected))
    )

    # Ensure axes is always treated as a list/iterable even if len(selected) == 1
    if len(selected) == 1:
        axes = [axes]

    for ax, (target, xml_file) in zip(axes, selected):
        root = ET.parse(xml_file).getroot()
        filename = root.findtext("filename")
        img_path = image_dir / filename
        
        image = cv2.imread(str(img_path))

        if image is None:
            ax.set_title(f"Image missing: {filename}")
            ax.axis("off")
            continue

        image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
        ax.imshow(image)

        for obj in root.findall("object"):
            name = obj.findtext("name", default="Unknown").strip()
            box = obj.find("bndbox")

            if box is None:
                continue

            x1 = float(box.findtext("xmin"))
            y1 = float(box.findtext("ymin"))
            x2 = float(box.findtext("xmax"))
            y2 = float(box.findtext("ymax"))

            color = "red" if name == target else "yellow"

            ax.add_patch(patches.Rectangle(
                (x1, y1), x2 - x1, y2 - y1,
                linewidth=2, edgecolor=color, facecolor="none"
            ))

            ax.text(
                x1, max(0, y1 - 5), name,
                color="white",
                bbox=dict(facecolor=color, alpha=0.8)
            )

        ax.set_title(f"Target class: {target} | {filename}")
        ax.axis("off")

    plt.tight_layout()
    plt.show()
```
